```python
import cv2 as cv
import numpy as np
from tensorflow.keras.applications.resnet50 import ResNet50, preprocess_input, decode_predictions

# [1] ResNet50 사전학습 모델 불러오기 (ImageNet 1000클래스)
model = ResNet50(weights='imagenet')

# [2] 이미지 읽기 (BGR)
img = cv.imread('rabbit.jpg')

# [3] 이미지 크기 224x224로 리사이즈 & 배치 차원 추가 (1, 224, 224, 3)
x = np.reshape(cv.resize(img, (224, 224)), (1, 224, 224, 3))

# [4] ResNet50 전처리 함수 (RGB로 변환 & 평균값 보정 등)
x = preprocess_input(x)

# [5] 모델 예측 수행 (벡터 형태 출력)
preds = model.predict(x)

# [6] ImageNet 클래스 이름으로 디코딩 (Top 5)
top5 = decode_predictions(preds, top=5)[0]
print('예측 결과:', top5)

# [7] 원본 이미지에 Top5 클래스 이름 + 신뢰도 출력
for i in range(5):
    label = top5[i][1]         # 클래스 이름
    confidence = str(top5[i][2])  # 신뢰도 (확률)
    cv.putText(
        img,
        label + ':' + confidence,  # ex: "rabbit:0.78"
        (10, 20 + i * 20),         # Y좌표 간격
        cv.FONT_HERSHEY_SIMPLEX,
        0.5,
        (255, 255, 255),
        1
    )

# [8] 결과 이미지 창 출력
cv.imshow('Recognition result', img)

cv.waitKey()
cv.destroyAllWindows()
```
